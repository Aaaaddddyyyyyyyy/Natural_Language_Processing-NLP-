```python
import spacy
```

```python
nlp=spacy.load('en_core_web_sm')      # importing english language dict
```

```python
doc=nlp(u'i will google about facebok')
```

```python
doc.text
```

```python
doc[0]
```

```python
doc[0].pos_
```

```python
for word in doc:
  print(word.text,'----->',word.pos_,word.tag_,spacy.explain(word.tag_))
```
