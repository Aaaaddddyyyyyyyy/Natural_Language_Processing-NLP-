```python
dummy= 'Data acquisition is not a one-time activity. As language evolves and user behavior changes, you need to continuously collect and update your data to maintain model performance.'
```

```python
dummy
```

```python
from nltk.tokenize import word_tokenize,sent_tokenize
```

```python
import nltk       # sentence tokenization
nltk.download('punkt_tab')
sents= sent_tokenize(dummy)
sents
```

```python
for sent in sents:              ## word tokenization
  print(word_tokenize(sent))
```
