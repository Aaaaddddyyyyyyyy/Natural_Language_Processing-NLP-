NLTK

```python
from nltk.tokenize import word_tokenize,sent_tokenize
```

```python
import nltk
nltk.download('punkt_tab')
sent1='i am going to delhi'
word_tokenize(sent1)
```

```python
# senetnce tokenization

sent1='iam gpoi g to visit delhi. suggest me some places to visit. i will stay here for 3 days'
sent_tokenize(sent1)
```
