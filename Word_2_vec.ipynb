```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
import os
import nltk

# Ensure Google Drive is mounted if not already done
# from google.colab import drive
# drive.mount('/content/drive')

# Download the 'punkt_tab' tokenizer data if not already downloaded
try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    nltk.download('punkt_tab')

story=[]
data_path = '/content/drive/MyDrive/data'
# It's good practice to ensure the directory exists before listing, though os.listdir will fail if it doesn't.
# os.makedirs(data_path, exist_ok=True) # Uncomment if you want to ensure the directory exists for writing files

for filename in os.listdir(data_path):

  f=open(os.path.join(data_path,filename), encoding='latin-1') # Corrected path and added encoding to open file from Google Drive
  corpus=f.read()
  raw_sent= sent_tokenize(corpus)

  for sent in raw_sent:
    story.append(simple_preprocess(sent))
```

```python
len(story)
```

MOdel building

```python
model=gensim.models.Word2Vec(
    window=10,
    min_count=2,
    workers=4,
)
```

```python
model.build_vocab(story)
```

```python
model.train(story,total_examples=model.corpus_count,epochs=model.epochs)
```
