```python
from nltk.stem.porter import PorterStemmer
```

```python
ps=PorterStemmer()
def stem_words(text):
  return" ".join([ps.stem(word) for word in text.split()])
```

```python
sample='walk walks walking walked'
stem_words(sample)
```

```python
text='She connected the devices and checked the connection.students are studying hard for their exams.'
stem_words(text)
```

LEmmatization

```python
import nltk
from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer=WordNetLemmatizer()
```

```python
from nltk.corpus.reader import wordnet
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
sent='  he was running and eating at the same time. he has bad habbit of swimming after playing long hours in sun.'
punctuations="?!.,:"
sent_words=nltk.word_tokenize(sent)

for word in sent_words:
  if word in punctuations:
    sent_words.remove(word)

sent_words
print("{0:20}{1:20}".format("Word","Lemma"))
for word in sent_words:
  print("{0:20}{1:20}".format(word,wordnet_lemmatizer.lemmatize(word)))
```
